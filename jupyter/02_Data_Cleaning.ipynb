# 4. Feature Engineering

## Objective

The `Date` column contains complete date information, but it is not very useful for business analysis in its current form.

In this step, we will extract useful features from the Date column, such as Year, Month, Day, Weekday, and Quarter.

These new features will help answer business questions like:

- Which month has the highest sales?
- Which weekday generates the most revenue?
- Which quarter performs the best?

These features will also improve the performance of our Machine Learning model.

```python
# Extract Year
data["Year"] = data["Date"].dt.year

# Extract Month
data["Month"] = data["Date"].dt.month

# Extract Day
data["Day"] = data["Date"].dt.day

# Extract Weekday
data["Weekday"] = data["Date"].dt.day_name()

# Extract Quarter
data["Quarter"] = data["Date"].dt.quarter
```

```python
data[["Date","Year","Month","Day","Weekday","Quarter"]].head()
```

## Observations

The Date column has been successfully transformed into multiple useful features.

These features will be used throughout the project for:

- Exploratory Data Analysis (EDA)
- SQL Business Queries
- Power BI Dashboards
- Machine Learning Models

Instead of working with a single Date column, we can now analyze sales trends by year, month, weekday, and quarter.
